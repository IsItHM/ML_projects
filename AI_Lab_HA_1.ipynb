<p style="font-size: 12px;"><b>Why .columns.tolist()?</b></p>

<p style="font-size: 12px;">The line has three steps:</p>

<ol>
<li style="font-size: 12px;"><code>X_train.select_dtypes(exclude='number')</code> returns a DataFrame with only the text columns, all rows included.</li>
<li style="font-size: 12px;"><code>.columns</code> keeps only the column names, as a pandas Index object.</li>
<li style="font-size: 12px;"><code>.tolist()</code> turns that Index into a plain Python list.</li>
</ol>

<p style="font-size: 12px;"><code>.tolist()</code> is not strictly required, because <code>X_train[cat_cols]</code> and ColumnTransformer also accept an Index. I used it for two reasons:</p>

<p style="font-size: 12px;"><b>1. The output is cleaner.</b> Here is what each one prints:</p>

<pre style="font-size: 12px;">
Index(['workclass', 'sex', ...], dtype='str')   # without .tolist()
['workclass', 'sex', ...]                       # with .tolist()
</pre>

<p style="font-size: 12px;"><b>2. The + operator behaves differently.</b> On lists, + joins them. On an Index, + adds the items pair by pair. For example, with cat_cols = ['workclass', 'sex'] and num_cols = ['age', 'hours']:</p>

<pre style="font-size: 12px;">
cat_cols + num_cols
# lists: ['workclass', 'sex', 'age', 'hours']   -> joined, as expected
# Index: Index(['workclassage', 'sexhours'])     -> names glued together pair by pair
</pre>

<p style="font-size: 12px;">With the real columns (7 categorical and 5 numerical), the Index version raises a ValueError because the lengths are different. Using lists avoids this problem.</p>

<h5><u>Task 20:</u>Apply an appropriate categorical encoding technique to convert categorical values into
numerical representations.</h5>

<p style="font-size: 12px;"><b>Why do we need pd.concat after encoding?</b></p>

<p style="font-size: 12px;">The encoder only sees the columns we pass to it. We passed <code>X_train[cat_cols]</code>, which has only the 7 categorical (text) columns, so the output contains only those, now as 86 one-hot columns. The numerical columns <code>age</code>, <code>education-num</code>, <code>capital-gain</code>, <code>capital-loss</code> and <code>hours-per-week</code> were never sent to the encoder, so they are missing from the result.</p>

<p style="font-size: 12px;">Example with row 24764 of the training set:</p>

<pre style="font-size: 12px;">
X_train (12 columns)
age | education-num | ... | workclass | ... | native-country
 33 |       9       | ... |  Private  | ... | United-States

X_train[cat_cols] -> encoder -> X_train_encoded (86 columns)
workclass_Federal-gov | workclass_Private | ... | native-country_United-States
         0.0          |        1.0        | ... |            1.0
-> age, education-num, etc. are gone
</pre>

<p style="font-size: 12px;">A model needs all the features, and age and hours worked are strong predictors of income. So the encoded columns and the numerical columns have to be joined back into one table:</p>

<pre style="font-size: 12px;">
pd.concat([X_train[num_cols], X_train_encoded], axis=1)

age | education-num | capital-gain | capital-loss | hours-per-week | workclass_Federal-gov | ... (91 columns)
 33 |       9       |      0       |      0       |       60       |          0.0          | ...
</pre>

<ul>
<li style="font-size: 12px;"><code>[X_train[num_cols], X_train_encoded]</code> are the two tables to join: 5 numerical columns and 86 encoded columns.</li>
<li style="font-size: 12px;"><code>axis=1</code> places them side by side (adds columns). <code>axis=0</code> would stack one table under the other (adds rows), which is wrong here.</li>
<li style="font-size: 12px;">Rows are matched by index, so row 24764's age goes next to row 24764's encoded values. That is why keeping the original index with <code>set_output(transform="pandas")</code> matters.</li>
</ul>

<p style="font-size: 12px;">For Task 20 itself these lines are optional, because the task only asks to encode the categorical values, and the encoder already does that. The concat just gives one complete table to look at. In Task 26, ColumnTransformer does this joining automatically.</p>

```python
from sklearn.preprocessing import OneHotEncoder

# set_output(transform="pandas") -> returns a DataFrame with column names instead of a plain array
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False).set_output(transform="pandas")

# Fit only on training categories
X_train_encoded = encoder.fit_transform(X_train[cat_cols])
X_test_encoded = encoder.transform(X_test[cat_cols])
```

<p style="font-size: 12px;">This cell below is not necessary for the assignment, but I included it because the model needs to get the output for all features, which is why I concatenated the numerical and categorical features after encoding.</p>

```python
# add the numerical columns back next to the encoded ones
X_train_encoded = pd.concat([X_train[num_cols], X_train_encoded], axis=1)
X_test_encoded = pd.concat([X_test[num_cols], X_test_encoded], axis=1)
```

<h5><u>Task 21:</u> Explain why your selected encoding technique is appropriate for the dataset.</h5>

<p style="font-size: 12px;"><b>Why I chose One-Hot Encoding</b></p>

<p style="font-size: 12px;">The 7 categorical features are <code>workclass</code>, <code>maritual-status</code>, <code>occupation</code>, <code>relationship</code>, <code>race</code>, <code>sex</code> and <code>native-country</code>. All of them are <b>nominal</b>: their categories are just names with no natural order. (<code>education</code> was the only ordered category, but I dropped it in Task 10 because <code>education-num</code> already stores the same information as numbers in the right order.)</p>

<p style="font-size: 12px;"><b>1. One-hot encoding does not create a fake order.</b> A label/ordinal encoder gives each category a number in alphabetical order. For <code>workclass</code> in the training set that would be:</p>

<pre style="font-size: 12px;">
Federal-gov      = 0
Local-gov        = 1
Never-worked     = 2
Private          = 3
Self-emp-inc     = 4
Self-emp-not-inc = 5
State-gov        = 6
Unknown          = 7
Without-pay      = 8
</pre>

<p style="font-size: 12px;">A model would treat these as real quantities. Without-pay (8) would count as "more" than Private (3), and Never-worked (2) would sit closer to Federal-gov (0) than to Without-pay (8), even though Never-worked and Without-pay are the two most similar groups. These numbers are only alphabetical positions, so this order means nothing. One-hot encoding gives every category its own 0/1 column instead, so no category is bigger or closer than another. Two real rows from the training set:</p>

<pre style="font-size: 12px;">
row     workclass            workclass_Federal-gov | ... | workclass_Private | workclass_Self-emp-inc | workclass_Self-emp-not-inc | ... | workclass_Without-pay
24764   Private           ->           0           | ... |         1         |           0            |             0              | ... |           0
2754    Self-emp-not-inc  ->           0           | ... |         0         |           0            |             1              | ... |           0
</pre>

<p style="font-size: 12px;"><b>2. The number of new columns is manageable.</b> The 7 features have 86 categories in total, so the data goes from 12 to 91 columns. With 26,029 training rows this is small. The largest feature is <code>native-country</code> with 41 categories, which is still fine.</p>

<p style="font-size: 12px;"><b>3. It handles unseen categories safely.</b> With <code>handle_unknown="ignore"</code>, a category that appears only in the test set becomes all 0s instead of causing an error.</p>

<p style="font-size: 12px;"><b>4. It avoids data leakage.</b> The encoder is fitted only on <code>X_train</code>, so it learns the list of categories from the training data and applies the same list to <code>X_test</code>.</p>

<p style="font-size: 12px;">one-hot encoding adds many columns, and most of their values are 0. For features with hundreds or thousands of categories this would be a problem, and another method (for example grouping rare categories together) would be better. This dataset has no such features, so one-hot encoding is the right choice.</p>

<p style="font-size: 12px;"><b>StandardScaler vs MinMaxScaler</b></p>

<pre style="font-size: 12px;">
StandardScaler:  z  = (x - mean) / std          -> mean 0, std 1, no fixed range
MinMaxScaler:    x' = (x - min) / (max - min)   -> between 0 and 1 (on the training data)
</pre>

<p style="font-size: 12px;"><b>1. Range.</b> MinMaxScaler puts every feature exactly between 0 and 1. StandardScaler has no fixed range: <code>age</code> goes from -1.58 to 3.77, and <code>capital-gain</code> from -0.15 to 13.55.</p>

<p style="font-size: 12px;"><b>2. Meaning of the values.</b> After StandardScaler every feature has mean 0, so a negative value means below average and a positive value means above average. After MinMaxScaler, 0 means the smallest value in the training set and 1 the largest, and the means are different for each feature (0.01 to 0.61).</p>

<p style="font-size: 12px;"><b>3. Shape stays the same.</b> Both scalers only subtract and divide, so the order of the rows and the shape of each distribution do not change. A skewed feature stays skewed with either scaler.</p>

<p style="font-size: 12px;"><b>4. Outliers.</b> Most people have <code>capital-gain</code> = 0, but the largest value is 99999. MinMaxScaler uses this maximum as the top of the scale, so most values are squeezed close to 0: after MinMax, <code>capital-gain</code> has mean 0.01 and std 0.07, while <code>age</code> has std 0.19. So the features still do not have equal spread. After StandardScaler, every feature has std 1.</p>

<p style="font-size: 12px;"><b>5. Test data.</b> Both scalers use the training statistics on the test set. With MinMaxScaler, a test value smaller than the training min or bigger than the training max would go below 0 or above 1. In this split the test set has the same min and max as the training set, so it stays between 0 and 1, but that is not guaranteed in general.</p>

<p style="font-size: 12px;"><b>Which one I will use:</b> StandardScaler. Because of the outliers in <code>capital-gain</code> and <code>capital-loss</code>, MinMaxScaler squeezes most of their values close to 0, while StandardScaler gives every feature the same spread (std 1). MinMaxScaler suits features with a natural fixed range and no extreme values, such as image pixels (0-255). Neither scaler fixes the skew. A log transform (<code>np.log1p</code>) or <code>RobustScaler</code> could help with that, but it is beyond this task.</p>

<h5><u>Task 25:</u> Explain why feature scaling is necessary and whether all machine learning algorithms
require it.</h5>

<p style="font-size: 12px;"><b>Why is feature scaling necessary?</b></p>

<p style="font-size: 12px;">In this dataset the numerical columns have very different ranges. For example, <code>education-num</code> only goes from 1 to 16, but <code>capital-gain</code> goes from 0 to 99999. Some algorithms only look at the numbers, so they think a column with bigger numbers is more important, even when it is not. Scaling brings all columns to a similar range so every column gets a fair chance.</p>

<p style="font-size: 12px;"><b>Do all algorithms need scaling?</b></p>

<p style="font-size: 12px;">No. Algorithms that need it:</p>
<ul>
<li style="font-size: 12px;">the algorithms that use distances.</li>
<li style="font-size: 12px;">Logistic regression and neural networks, because they learn step by step and different ranges make learning slow.</li>
</ul>

<p style="font-size: 12px;">Algorithms that don't need it: Decision Tree and Random Forest. They ask simple questions about one column at a time, like "is age > 40?". After scaling, the number in the question changes, but the same people still answer yes or no, so the result is the same.</p>

```python
# now reeloading the data before imputation, so the pipeline has real NaNs to fill
df2 = pd.read_csv(url, names=coloumn_name, skipinitialspace=True, na_values='?')
df2 = df2.drop_duplicates().drop(columns=['fnlwgt', 'education'])

X2 = df2.drop(columns=['class'])
y2 = df2['class'].map({'<=50K': 0, '>50K': 1})

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42, stratify=y2)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

numeric_features = ['age', 'education-num', 'capital-gain', 'capital-loss', 'hours-per-week']
categorical_features = ['workclass', 'maritual-status', 'occupation', 'relationship',
                        'race', 'sex', 'native-country']

# Numerical: fill missing with median, then standardize
numeric_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical: fill missing with most frequent value, then one-hot encode
categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Send each column group to its own pipeline
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_pipeline, numeric_features),
    ('cat', categorical_pipeline, categorical_features)
])
```

```python
X2_train_prep = preprocessor.fit_transform(X2_train)
X2_test_prep  = preprocessor.transform(X2_test)
```
